```python
import pandas as pd
from openpyxl import Workbook
from openpyxl.styles import Font, Alignment
from openpyxl.chart import LineChart, Reference

data = {
2010:1289, 2011:1305, 2012:1349, 2013:1365, 2014:1368, 2015:1428,
2016:1443, 2017:1465, 2018:1539, 2019:1579, 2020:1633, 2021:1660,
2022:1789, 2023:1890, 2024:1921, 2025:2015
}

rows = []
for year, weekly in data.items():
    hourly = weekly / 40
    rows.append([year, weekly, hourly, weekly * 52, hourly * 2080])

df = pd.DataFrame(rows, columns=[
    "Year", "Median Weekly Earnings ($)", "Estimated Median Hourly Wage ($)",
    "Annualized Weekly Earnings ($)", "Annualized Hourly Wage ($)"
])

path = "/mnt/data/US_CS_Median_Wage_2010_2025.xlsx"
with pd.ExcelWriter(path, engine="openpyxl") as writer:
    df.to_excel(writer, index=False, sheet_name="CS Median Wage")
    ws = writer.book["CS Median Wage"]
    ws.freeze_panes = "A2"
    for cell in ws[1]:
        cell.font = Font(bold=True)
        cell.alignment = Alignment(horizontal="center")
    for col in ws.columns:
        width = max(len(str(c.value)) if c.value is not None else 0 for c in col) + 2
        ws.column_dimensions[col[0].column_letter].width = min(width, 32)
    for row in ws.iter_rows(min_row=2, min_col=2, max_col=5):
        for cell in row:
            cell.number_format = '$#,##0.00'
    chart = LineChart()
    chart.title = "US Computer & Mathematical Occupations – Median Wage"
    chart.y_axis.title = "Median Hourly Wage ($)"
    chart.x_axis.title = "Year"
    chart.add_data(Reference(ws, min_col=3, min_row=1, max_row=len(df)+1), titles_from_data=True)
    chart.set_categories(Reference(ws, min_col=1, min_row=2, max_row=len(df)+1))
    chart.height = 8
    chart.width = 15
    ws.add_chart(chart, "G2")

print(path)
```
